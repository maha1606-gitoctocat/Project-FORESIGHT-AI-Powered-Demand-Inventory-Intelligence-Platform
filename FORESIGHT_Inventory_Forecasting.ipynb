```python
weekly_sales = (
    sales
    .groupby([
        pd.Grouper(key='date', freq='W-MON'),
        'sku_id'
    ])['quantity']
    .sum()
    .reset_index()
)

weekly_sales.rename(columns={'quantity': 'weekly_demand'}, inplace=True)

weekly_sales.head()
```

```python
# Create a 52-week seasonal-naive baseline

weekly_sales = weekly_sales.sort_values(['sku_id', 'date']).copy()

weekly_sales['seasonal_naive'] = (
    weekly_sales.groupby('sku_id')['weekly_demand']
    .shift(52)
)

weekly_sales.head(60)
```

```python
# Calculate WAPE for the seasonal-naive baseline

baseline_eval = weekly_sales.dropna(subset=['seasonal_naive']).copy()

wape_baseline = (
    (baseline_eval['weekly_demand'] - baseline_eval['seasonal_naive']).abs().sum()
    / baseline_eval['weekly_demand'].abs().sum()
) * 100

print(f"Seasonal-Naive Baseline WAPE: {wape_baseline:.2f}%")
```

```python
# Create time-based features

weekly_sales['year'] = weekly_sales['date'].dt.year
weekly_sales['month'] = weekly_sales['date'].dt.month
weekly_sales['week_of_year'] = weekly_sales['date'].dt.isocalendar().week.astype(int)

weekly_sales['trend'] = (
    weekly_sales.groupby('sku_id').cumcount()
)

weekly_sales[['date', 'sku_id', 'weekly_demand',
              'year', 'month', 'week_of_year', 'trend']].head()
```

```python
# Create lag features for each SKU

for lag in [1, 4, 12, 52]:
    weekly_sales[f'lag_{lag}'] = (
        weekly_sales.groupby('sku_id')['weekly_demand']
        .shift(lag)
    )

weekly_sales[
    ['date', 'sku_id', 'weekly_demand',
     'lag_1', 'lag_4', 'lag_12', 'lag_52']
].head(60)
```

```python
# Create rolling demand features

weekly_sales['rolling_mean_4'] = (
    weekly_sales.groupby('sku_id')['weekly_demand']
    .transform(lambda x: x.shift(1).rolling(4).mean())
)

weekly_sales['rolling_mean_12'] = (
    weekly_sales.groupby('sku_id')['weekly_demand']
    .transform(lambda x: x.shift(1).rolling(12).mean())
)

weekly_sales[
    ['date', 'sku_id', 'weekly_demand',
     'rolling_mean_4', 'rolling_mean_12']
].head(20)
```

```python
# Prepare data for model training

model_data = weekly_sales.dropna(
    subset=[
        'lag_1',
        'lag_4',
        'lag_12',
        'lag_52',
        'rolling_mean_4',
        'rolling_mean_12'
    ]
).copy()

print("Rows available for modeling:", len(model_data))
print("SKUs available:", model_data['sku_id'].nunique())
print("Start date:", model_data['date'].min())
print("End date:", model_data['date'].max())
```

```python
# Time-based train/test split

test_weeks = 8

cutoff_date = model_data['date'].max() - pd.Timedelta(weeks=test_weeks - 1)

train_data = model_data[
    model_data['date'] < cutoff_date
].copy()

test_data = model_data[
    model_data['date'] >= cutoff_date
].copy()

print("Train start:", train_data['date'].min())
print("Train end:", train_data['date'].max())
print("Test start:", test_data['date'].min())
print("Test end:", test_data['date'].max())
print("Train rows:", len(train_data))
print("Test rows:", len(test_data))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

# Features used by the model
features = [
    'lag_1',
    'lag_4',
    'lag_12',
    'lag_52',
    'rolling_mean_4',
    'rolling_mean_12',
    'month',
    'week_of_year',
    'trend'
]

X_train = train_data[features]
y_train = train_data['weekly_demand']

X_test = test_data[features]
y_test = test_data['weekly_demand']

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
```

```python
# Train Random Forest forecasting model

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

print("Training model...")

rf_model.fit(X_train, y_train)

print("Model training completed!")
```

```python
# Generate predictions on the test data

y_pred = rf_model.predict(X_test)

# Make sure predictions cannot be negative
y_pred = np.maximum(y_pred, 0)

print("Predictions generated:", len(y_pred))
print("First 10 predictions:")
print(y_pred[:10])
```

```python
# Calculate WAPE for the Random Forest model

wape_rf = (
    np.abs(y_test - y_pred).sum()
    / np.abs(y_test).sum()
) * 100

print(f"Random Forest WAPE: {wape_rf:.2f}%")
print(f"Seasonal-Naive WAPE: {wape_baseline:.2f}%")

if wape_rf < wape_baseline:
    print("Random Forest performs better than the seasonal-naive baseline.")
else:
    print("Seasonal-naive baseline performs better than Random Forest.")
```

```python
# Define rolling-origin backtest dates

backtest_dates = sorted(backtest_data['date'].unique())

fold_dates = [
    backtest_dates[-24],
    backtest_dates[-16],
    backtest_dates[-8]
]

print("Fold 1 test date:", fold_dates[0])
print("Fold 2 test date:", fold_dates[1])
print("Fold 3 test date:", fold_dates[2])
```

```python
# Rolling-origin backtest

fold_results = []

for i, test_start in enumerate(fold_dates, start=1):

    # Use 8 weeks as the test window
    test_end = test_start + pd.Timedelta(weeks=7)

    fold_train = backtest_data[
        backtest_data['date'] < test_start
    ].copy()

    fold_test = backtest_data[
        (backtest_data['date'] >= test_start) &
        (backtest_data['date'] <= test_end)
    ].copy()

    X_fold_train = fold_train[features]
    y_fold_train = fold_train['weekly_demand']

    X_fold_test = fold_test[features]
    y_fold_test = fold_test['weekly_demand']

    fold_model = RandomForestRegressor(
        n_estimators=50,
        max_depth=15,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    )

    print(f"Training Fold {i}...")

    fold_model.fit(X_fold_train, y_fold_train)

    fold_pred = fold_model.predict(X_fold_test)
    fold_pred = np.maximum(fold_pred, 0)

    fold_wape = (
        np.abs(y_fold_test - fold_pred).sum()
        / np.abs(y_fold_test).sum()
    ) * 100

    fold_results.append({
        'fold': i,
        'test_start': test_start,
        'test_end': test_end,
        'wape': fold_wape,
        'test_rows': len(fold_test)
    })

    print(f"Fold {i} WAPE: {fold_wape:.2f}%")

fold_results_df = pd.DataFrame(fold_results)

print("\nRolling-origin results:")
print(fold_results_df)
```

```python
# Correct Seasonal Naive rolling-origin backtest

naive_fold_results = []

# Use the complete historical dataset so 52-week history is available
naive_data = weekly_sales.copy()

for i, test_start in enumerate(fold_dates, start=1):

    # Same 8-week test window as Random Forest
    test_end = test_start + pd.Timedelta(weeks=7)

    # Select the 8-week test period
    fold_test = naive_data[
        (naive_data['date'] >= test_start) &
        (naive_data['date'] <= test_end)
    ].copy()

    # Seasonal Naive:
    # prediction = demand from the same SKU 52 weeks earlier
    naive_data_sorted = naive_data.sort_values(
        ['sku_id', 'date']
    ).copy()

    naive_data_sorted['naive_pred'] = (
        naive_data_sorted
        .groupby('sku_id')['weekly_demand']
        .shift(52)
    )

    # Get predictions for this fold's test dates
    fold_test = naive_data_sorted[
        (naive_data_sorted['date'] >= test_start) &
        (naive_data_sorted['date'] <= test_end)
    ].copy()

    # Keep only rows where 52-week historical demand exists
    valid_test = fold_test.dropna(
        subset=['naive_pred']
    ).copy()

    y_actual = valid_test['weekly_demand']
    y_naive = valid_test['naive_pred']

    naive_wape = (
        np.abs(y_actual - y_naive).sum()
        / np.abs(y_actual).sum()
    ) * 100

    naive_fold_results.append({
        'fold': i,
        'test_start': test_start,
        'test_end': test_end,
        'naive_wape': naive_wape,
        'test_rows': len(valid_test)
    })

    print(
        f"Naive Fold {i} WAPE: "
        f"{naive_wape:.2f}%"
    )

naive_fold_results_df = pd.DataFrame(
    naive_fold_results
)

print("\nSeasonal Naive rolling-origin results:")
print(naive_fold_results_df)

average_naive_wape = (
    naive_fold_results_df['naive_wape'].mean()
)

print(
    f"\nAverage Seasonal Naive WAPE: "
    f"{average_naive_wape:.2f}%"
)
```

```python
# Create future 8-week forecast

import pandas as pd
import numpy as np

# Number of future weeks
forecast_horizon = 8

# Make a copy of historical data
future_base = weekly_sales[
    ['date', 'sku_id', 'weekly_demand']
].copy()

future_base = future_base.sort_values(
    ['sku_id', 'date']
).reset_index(drop=True)

# Store the latest date
last_date = future_base['date'].max()

# Get all SKUs
sku_list = future_base['sku_id'].unique()

# Create future dates
future_dates = pd.date_range(
    start=last_date + pd.Timedelta(weeks=1),
    periods=forecast_horizon,
    freq='7D'
)

# Create SKU × future-date combinations
future_rows = pd.MultiIndex.from_product(
    [sku_list, future_dates],
    names=['sku_id', 'date']
).to_frame(index=False)

# Combine historical and future rows
forecast_history = future_base.copy()
forecast_history['is_future'] = False

future_rows['weekly_demand'] = np.nan
future_rows['is_future'] = True

forecast_history = pd.concat(
    [forecast_history, future_rows],
    ignore_index=True
)

forecast_history = forecast_history.sort_values(
    ['sku_id', 'date']
).reset_index(drop=True)

# Create time features
forecast_history['year'] = forecast_history['date'].dt.year
forecast_history['month'] = forecast_history['date'].dt.month
forecast_history['week_of_year'] = (
    forecast_history['date']
    .dt.isocalendar()
    .week
    .astype(int)
)

# Trend continues from historical data
forecast_history['trend'] = (
    forecast_history.groupby('sku_id').cumcount()
)

# Generate forecasts recursively
future_predictions = []

for sku in sku_list:

    sku_data = forecast_history[
        forecast_history['sku_id'] == sku
    ].copy()

    for idx in sku_data.index:

        if not sku_data.loc[idx, 'is_future']:
            continue

        # Create lag features using previous predictions/actuals
        previous_values = sku_data.loc[
            sku_data.index < idx,
            'weekly_demand'
        ].values

        if len(previous_values) < 52:
            continue

        sku_data.loc[idx, 'lag_1'] = previous_values[-1]
        sku_data.loc[idx, 'lag_4'] = previous_values[-4]
        sku_data.loc[idx, 'lag_12'] = previous_values[-12]
        sku_data.loc[idx, 'lag_52'] = previous_values[-52]

        sku_data.loc[idx, 'rolling_mean_4'] = (
            np.mean(previous_values[-4:])
        )

        sku_data.loc[idx, 'rolling_mean_12'] = (
            np.mean(previous_values[-12:])
        )

        # Prepare model input
        X_future = sku_data.loc[
            [idx],
            features
        ]

        # Predict
        prediction = rf_model.predict(X_future)[0]

        # Demand cannot be negative
        prediction = max(prediction, 0)

        # Store prediction so the next future week can use it
        sku_data.loc[idx, 'weekly_demand'] = prediction

        future_predictions.append({
            'date': sku_data.loc[idx, 'date'],
            'sku_id': sku,
            'forecast_demand': prediction
        })

# Convert predictions to DataFrame
future_forecast = pd.DataFrame(
    future_predictions
)

print("Future forecast created successfully.")
print("Forecast start:", future_forecast['date'].min())
print("Forecast end:", future_forecast['date'].max())
print("SKUs forecasted:", future_forecast['sku_id'].nunique())
print("Rows:", len(future_forecast))

future_forecast.head(10)
```

```python
# Merge 8-week forecast with store-level inventory

risk_forecast = risk_inventory.merge(
    sku_future_demand,
    on='sku_id',
    how='left'
)

print("Risk forecast shape:", risk_forecast.shape)

print("\nMissing future forecasts:",
      risk_forecast['forecast_8_week_demand'].isna().sum())

print("\nSample:")
display(
    risk_forecast[
        [
            'store_id',
            'sku_id',
            'stock_on_hand',
            'reorder_point',
            'safety_stock',
            'cost_price',
            'forecast_8_week_demand'
        ]
    ].head(10)
)
```

```python
# Calculate stock coverage against 8-week forecast

risk_forecast['forecast_8_week_demand'] = (
    risk_forecast['forecast_8_week_demand']
    .clip(lower=0)
)

risk_forecast['stock_coverage_ratio'] = np.where(
    risk_forecast['forecast_8_week_demand'] > 0,
    risk_forecast['stock_on_hand']
    / risk_forecast['forecast_8_week_demand'],
    np.inf
)

risk_forecast['forecast_stock_gap'] = (
    risk_forecast['stock_on_hand']
    - risk_forecast['forecast_8_week_demand']
)

print("Stock coverage calculated.")

display(
    risk_forecast[
        [
            'store_id',
            'sku_id',
            'stock_on_hand',
            'forecast_8_week_demand',
            'stock_coverage_ratio',
            'forecast_stock_gap'
        ]
    ].head(10)
)
```

```python
# Check available inventory columns for lead-time information

print("Inventory columns:")
print(inventory.columns.tolist())

print("\nRisk inventory columns:")
print(risk_inventory.columns.tolist())
```

```python
# Forecast-driven stockout risk

risk_forecast['forecast_stockout_risk'] = np.select(
    [
        risk_forecast['stock_on_hand'] < risk_forecast['forecast_8_week_demand'],
        
        (
            (risk_forecast['stock_on_hand'] >= risk_forecast['forecast_8_week_demand']) &
            (
                (risk_forecast['stock_on_hand'] <= risk_forecast['reorder_point']) |
                (risk_forecast['stock_on_hand'] <= risk_forecast['safety_stock'])
            )
        )
    ],
    [
        'HIGH',
        'MEDIUM'
    ],
    default='LOW'
)

print("Forecast-driven stockout risk created.")

print("\nRisk distribution:")
print(
    risk_forecast['forecast_stockout_risk']
    .value_counts()
)

display(
    risk_forecast[
        [
            'store_id',
            'sku_id',
            'stock_on_hand',
            'forecast_8_week_demand',
            'reorder_point',
            'safety_stock',
            'forecast_stockout_risk'
        ]
    ].head(10)
)
```

```python
# Forecast-driven overstock risk

risk_forecast['forecast_overstock_risk'] = np.select(
    [
        risk_forecast['stock_coverage_ratio'] >= 2,
        risk_forecast['stock_coverage_ratio'] >= 1.5
    ],
    [
        'HIGH',
        'MEDIUM'
    ],
    default='LOW'
)

print("Forecast-driven overstock risk created.")

print("\nOverstock risk distribution:")
print(
    risk_forecast['forecast_overstock_risk']
    .value_counts()
)

display(
    risk_forecast[
        [
            'store_id',
            'sku_id',
            'stock_on_hand',
            'forecast_8_week_demand',
            'stock_coverage_ratio',
            'forecast_overstock_risk'
        ]
    ].head(10)
)
```

```python
# Create recommended inventory actions

risk_forecast['recommended_action'] = np.select(
    [
        risk_forecast['forecast_stockout_risk'] == 'HIGH',
        risk_forecast['forecast_overstock_risk'] == 'HIGH',
        risk_forecast['forecast_stockout_risk'] == 'MEDIUM',
        risk_forecast['forecast_overstock_risk'] == 'MEDIUM'
    ],
    [
        'Reorder Now',
        'Clear / Markdown',
        'Watch / Replenish Soon',
        'Watch / Reduce Orders'
    ],
    default='Healthy'
)

print("Recommended actions created.")

print("\nAction distribution:")
print(
    risk_forecast['recommended_action']
    .value_counts()
)

display(
    risk_forecast[
        [
            'store_id',
            'sku_id',
            'stock_on_hand',
            'forecast_8_week_demand',
            'forecast_stockout_risk',
            'forecast_overstock_risk',
            'recommended_action'
        ]
    ].head(10)
)
```

```python
# Calculate inventory value at stake

risk_forecast['stockout_units_at_risk'] = (
    risk_forecast['forecast_8_week_demand']
    - risk_forecast['stock_on_hand']
).clip(lower=0)

risk_forecast['overstock_units_at_risk'] = (
    risk_forecast['stock_on_hand']
    - risk_forecast['forecast_8_week_demand']
).clip(lower=0)

# Convert unit exposure into rupee value
risk_forecast['stockout_value_at_risk'] = (
    risk_forecast['stockout_units_at_risk']
    * risk_forecast['cost_price']
)

risk_forecast['overstock_value_at_risk'] = (
    risk_forecast['overstock_units_at_risk']
    * risk_forecast['cost_price']
)

# Total value at stake
risk_forecast['total_value_at_stake'] = (
    risk_forecast['stockout_value_at_risk']
    + risk_forecast['overstock_value_at_risk']
)

print("Value-at-stake calculations completed.")

print("\nTotal stockout value at risk: ₹",
      f"{risk_forecast['stockout_value_at_risk'].sum():,.2f}")

print("Total overstock value at risk: ₹",
      f"{risk_forecast['overstock_value_at_risk'].sum():,.2f}")

print("Total value at stake: ₹",
      f"{risk_forecast['total_value_at_stake'].sum():,.2f}")

display(
    risk_forecast[
        [
            'store_id',
            'sku_id',
            'stock_on_hand',
            'forecast_8_week_demand',
            'cost_price',
            'stockout_units_at_risk',
            'overstock_units_at_risk',
            'stockout_value_at_risk',
            'overstock_value_at_risk',
            'total_value_at_stake'
        ]
    ].head(10)
)
```

```python
# Create final inventory risk output

final_risk_output = risk_forecast[
    [
        'store_id',
        'sku_id',
        'category',
        'subcategory',
        'brand',
        'stock_on_hand',
        'reorder_point',
        'safety_stock',
        'forecast_8_week_demand',
        'stock_coverage_ratio',
        'forecast_stockout_risk',
        'forecast_overstock_risk',
        'recommended_action',
        'cost_price',
        'stockout_units_at_risk',
        'overstock_units_at_risk',
        'stockout_value_at_risk',
        'overstock_value_at_risk',
        'total_value_at_stake'
    ]
].copy()

print("Final risk output created.")
print("Rows:", len(final_risk_output))
print("Columns:", len(final_risk_output.columns))

display(final_risk_output.head(10))
```

```python
# Finalize value-at-stake based on risk classification

# Stockout exposure:
# Count shortage value only for MEDIUM/HIGH stockout risk
risk_forecast['final_stockout_value_at_risk'] = np.where(
    risk_forecast['forecast_stockout_risk'].isin(['HIGH', 'MEDIUM']),
    risk_forecast['stockout_units_at_risk'] * risk_forecast['cost_price'],
    0
)

# Overstock exposure:
# Count excess inventory value only for MEDIUM/HIGH overstock risk
risk_forecast['final_overstock_value_at_risk'] = np.where(
    risk_forecast['forecast_overstock_risk'].isin(['HIGH', 'MEDIUM']),
    risk_forecast['overstock_units_at_risk'] * risk_forecast['cost_price'],
    0
)

# Combined financial exposure
risk_forecast['final_value_at_stake'] = (
    risk_forecast['final_stockout_value_at_risk']
    + risk_forecast['final_overstock_value_at_risk']
)

print("Final value-at-stake calculation completed.")

print("\nFinal stockout value at risk: ₹",
      f"{risk_forecast['final_stockout_value_at_risk'].sum():,.2f}")

print("Final overstock value at risk: ₹",
      f"{risk_forecast['final_overstock_value_at_risk'].sum():,.2f}")

print("Final total value at stake: ₹",
      f"{risk_forecast['final_value_at_stake'].sum():,.2f}")
```

```python
# Create final risk output with corrected financial exposure

final_risk_output = risk_forecast[
    [
        'store_id',
        'sku_id',
        'category',
        'subcategory',
        'brand',
        'stock_on_hand',
        'reorder_point',
        'safety_stock',
        'forecast_8_week_demand',
        'stock_coverage_ratio',
        'forecast_stockout_risk',
        'forecast_overstock_risk',
        'recommended_action',
        'cost_price',
        'stockout_units_at_risk',
        'overstock_units_at_risk'
    ]
].copy()

# Add corrected financial values
final_risk_output['stockout_value_at_risk'] = (
    risk_forecast['final_stockout_value_at_risk'].values
)

final_risk_output['overstock_value_at_risk'] = (
    risk_forecast['final_overstock_value_at_risk'].values
)

final_risk_output['total_value_at_stake'] = (
    risk_forecast['final_value_at_stake'].values
)

print("Final risk output updated.")
print("Rows:", len(final_risk_output))
print("Columns:", len(final_risk_output.columns))

display(final_risk_output.head(10))
```

```python
# Save final inventory risk output

final_risk_output.to_csv(
    'final_inventory_risk_output.csv',
    index=False
)

print("Final inventory risk file saved successfully.")
print("Rows saved:", len(final_risk_output))
print("Columns saved:", len(final_risk_output.columns))
```
